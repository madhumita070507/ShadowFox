```python
import tensorflow as tf
from transformers import TFGPT2LMHeadModel, GPT2Tokenizer
import numpy as np
import matplotlib.pyplot as plt
from nltk.translate.bleu_score import sentence_bleu

# Load GPT-2 model and tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = TFGPT2LMHeadModel.from_pretrained("gpt2", pad_token_id=tokenizer.eos_token_id)

# Function to calculate perplexity
def calculate_perplexity(text):
    input_ids = tokenizer.encode(text, return_tensors='tf')
    outputs = model(input_ids, labels=input_ids)
    loss = outputs.loss
    return np.exp(loss.numpy())

# Function to calculate BLEU score
def calculate_bleu(generated_text, reference_text):
    generated_tokens = generated_text.split()
    reference_tokens = [ref.split() for ref in reference_text]
    return sentence_bleu(reference_tokens, generated_tokens)

# Function to generate text and evaluate
def generate_text_and_evaluate(prompt, reference):
    input_ids = tokenizer.encode(prompt, return_tensors='tf')
    output_ids = model.generate(
        input_ids,
        max_length=100,
        num_beams=5,
        no_repeat_ngram_size=2,
        early_stopping=True
    )
    generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
    ppl = calculate_perplexity(generated_text)
    bleu = calculate_bleu(generated_text, reference)
    return generated_text, ppl, bleu

# Test prompts
test_prompts = [
    "Once upon a time",
    "The future of AI is",
    "In a distant galaxy"
]

reference = ["Once upon a time, there was a small village."]

# Collect results
generated_texts = []
perplexities = []
bleu_scores = []

for prompt in test_prompts:
    gen_text, ppl, bleu = generate_text_and_evaluate(prompt, reference)
    generated_texts.append(gen_text)
    perplexities.append(float(ppl))  # ✅ FIX applied
    bleu_scores.append(bleu)

# Print results
for i in range(len(test_prompts)):
    print(f"\nPrompt: {test_prompts[i]}")
    print(f"Generated: {generated_texts[i]}")
    print(f"Perplexity: {perplexities[i]}")
    print(f"BLEU Score: {bleu_scores[i]}")

# Visualization
plt.figure(figsize=(12, 5))

# Perplexity plot
plt.subplot(1, 2, 1)
plt.bar(range(len(test_prompts)), perplexities, color='skyblue')
plt.xticks(range(len(test_prompts)), test_prompts, rotation=45)
plt.ylabel("Perplexity")
plt.title("Perplexity per Prompt")

# BLEU score plot
plt.subplot(1, 2, 2)
plt.bar(range(len(test_prompts)), bleu_scores, color='salmon')
plt.xticks(range(len(test_prompts)), test_prompts, rotation=45)
plt.ylabel("BLEU Score")
plt.title("BLEU Score per Prompt")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from nltk.sentiment import SentimentIntensityAnalyzer
import numpy as np
import nltk

# Download the necessary NLTK resources
nltk.download('vader_lexicon')

# Initialize the GPT-2 model and tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2", pad_token_id=tokenizer.eos_token_id)

# Function to calculate perplexity
def calculate_perplexity(text):
    inputs = tokenizer.encode(text, return_tensors="pt")
    with torch.no_grad():
        outputs = model(input_ids=inputs, labels=inputs)
        loss = outputs.loss
    perplexity = torch.exp(loss)
    return perplexity.item()

# Example sentences to evaluate
sentences = [
    "The weather is beautiful today!",
    "I am learning to use GPT-2 for text generation.",
    "Artificial intelligence is a fascinating field.",
    "Text generation models are changing how we write."
]

# Calculate perplexity for each sentence
perplexity_scores = [calculate_perplexity(sentence) for sentence in sentences]

# Function to calculate BLEU score with smoothing
def calculate_bleu(reference, generated):
    smoothing_function = SmoothingFunction().method1  # Apply smoothing
    return sentence_bleu([reference], generated, smoothing_function=smoothing_function)

# Reference sentence (this is just an example, you can add more references)
reference_sentence = ["The", "weather", "is", "beautiful", "today"]

# Calculate BLEU score for each generated sentence
bleu_scores = [calculate_bleu(reference_sentence, sentence.split()) for sentence in sentences]

# Initialize the VADER sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Function to analyze sentiment
def analyze_sentiment(text):
    sentiment = sia.polarity_scores(text)
    return sentiment

# Analyze sentiment for each sentence
sentiments = [analyze_sentiment(sentence) for sentence in sentences]

# Extract positive, neutral, and negative scores
positive_scores = [sentiment['pos'] for sentiment in sentiments]
neutral_scores = [sentiment['neu'] for sentiment in sentiments]
negative_scores = [sentiment['neg'] for sentiment in sentiments]

# Plotting Perplexity, BLEU Score, and Sentiment Analysis in Subplots
fig, axs = plt.subplots(3, 1, figsize=(12, 18))

# Perplexity Plot
axs[0].bar(sentences, perplexity_scores, color='skyblue')
axs[0].set_title('Perplexity of Generated Texts')
axs[0].set_xlabel('Sentences')
axs[0].set_ylabel('Perplexity')
axs[0].tick_params(axis='x', rotation=45)

# BLEU Score Plot
axs[1].bar(sentences, bleu_scores, color='lightgreen')
axs[1].set_title('BLEU Scores of Generated Texts')
axs[1].set_xlabel('Sentences')
axs[1].set_ylabel('BLEU Score')
axs[1].tick_params(axis='x', rotation=45)

# Sentiment Analysis Plot
x = np.arange(len(sentences))  # positions for each bar
axs[2].bar(x - 0.2, positive_scores, 0.4, label='Positive', color='lightblue')
axs[2].bar(x + 0.2, neutral_scores, 0.4, label='Neutral', color='gray')
axs[2].bar(x + 0.6, negative_scores, 0.4, label='Negative', color='salmon')
axs[2].set_title('Sentiment Analysis of Generated Texts')
axs[2].set_xlabel('Sentences')
axs[2].set_ylabel('Sentiment Scores')
axs[2].tick_params(axis='x', rotation=45)
axs[2].legend()

plt.tight_layout()
plt.show()
```
